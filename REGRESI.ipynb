## Visualisasi regresi

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go

# 1. MEMBUAT DATASET
data = {
    'Partisipan': np.arange(1, 12),  # 11 data observasi
    'Ketimpangan': [0.41, 0.33, 0.36, 0.46, 0.35, 0.49, 0.36, 0.32, 0.48, 0.50, 0.35],
    'Pengangguran': [8.72, 5.72, 6.15, 6.72, 9.02, 3.97, 3.13, 5.42, 9.63, 6.76, 3.14],
    'Kemiskinan': [14.63, 15.47, 13.31, 16.15, 7.82, 16.04, 15.30, 18.99, 19.56, 11.97, 12.53]
}

df = pd.DataFrame(data)

# 2. MENENTUKAN VARIABEL INDEPENDEN & DEPENDEN
X = df[['Ketimpangan', 'Pengangguran']]
y = df['Kemiskinan']

# 3. MENAMBAHKAN INTERCEPT
X_model = sm.add_constant(X)

# 4. ESTIMASI MODEL OLS
model = sm.OLS(y, X_model)
results = model.fit()

# 5. MENAMPILKAN RINGKASAN HASIL REGRESI
print("\n" + "=" * 70)
print(" HASIL REGRESI LINEAR BERGANDA")
print("=" * 70)
print(results.summary())

# 6. MENGAMBIL KOEFISIEN REGRESI
intercept = results.params["const"]
coef_ketimpangan = results.params["Ketimpangan"]
coef_pengangguran = results.params["Pengangguran"]
r_squared = results.rsquared
adjusted_r_squared = results.rsquared_adj

# 7. MEMBUAT PERSAMAAN REGRESI
persamaan = (
    f"Ŷ = {intercept:.3f}"
    f" + ({coef_ketimpangan:.3f}) X1"
    f" + ({coef_pengangguran:.3f}) X2"
)
print("\n" + "=" * 70)
print(" PERSAMAAN REGRESI")
print("=" * 70)
print(persamaan)
print(f"\nR2 = {r_squared:.4f}")
print(f"Adjusted R2 = {adjusted_r_squared:.4f}")

# 8. MENGHITUNG NILAI PREDIKSI
df["Prediksi"] = results.predict(X_model)
print("\n" + "=" * 70)
print(" DATA AKTUAL & PREDIKSI")
print("=" * 70)
print(
    df[["Partisipan", "Ketimpangan", "Pengangguran", "Kemiskinan", "Prediksi"]]
    .round(2)
)

# 9. MEMBUAT GRID UNTUK BIDANG REGRESI 3D
x1 = np.linspace(df["Ketimpangan"].min(), df["Ketimpangan"].max(), 40)
x2 = np.linspace(df["Pengangguran"].min(), df["Pengangguran"].max(), 40)
X1, X2 = np.meshgrid(x1, x2)

# 10. MENGHITUNG NILAI Y PADA BIDANG REGRESI
Y = intercept + (coef_ketimpangan * X1) + (coef_pengangguran * X2)

# 11. MEMBUAT FIGURE 3D DENGAN PLOTLY
fig = go.Figure()

# 12. MENAMPILKAN DATA AKTUAL (TITIK MERAH)
fig.add_trace(
    go.Scatter3d(
        x=df["Ketimpangan"],
        y=df["Pengangguran"],
        z=df["Kemiskinan"],
        mode="markers",
        marker=dict(size=7, color="red", opacity=0.9),
        name="Data Aktual",
        text=[f"Observasi {p}" for p in df["Partisipan"]],
        hovertemplate=(
            "<b>%{text}</b><br>"
            "Ketimpangan: %{x}<br>"
            "Pengangguran: %{y}<br>"
            "Kemiskinan: %{z}"
            "<extra></extra>"
        ),
    )
)

# 13. MENAMPILKAN BIDANG REGRESI (SURFACE)
fig.add_trace(
    go.Surface(
        x=X1,
        y=X2,
        z=Y,
        opacity=0.60,
        colorscale="Viridis",
        name="Bidang Regresi",
        hovertemplate=(
            "Ketimpangan: %{x:.2f}<br>"
            "Pengangguran: %{y:.2f}<br>"
            "Prediksi Y: %{z:.2f}"
            "<extra></extra>"
        ),
    )
)

# 14. MENGATUR TAMPILAN GRAFIK
fig.update_layout(
    title={
        "text": f"REGRESI LINEAR BERGANDA 3D<br><sup>{persamaan} | R2 = {r_squared:.3f}</sup>",
        "x": 0.5,
    },
    scene={
        "xaxis": {"title": "Ketimpangan (X1)"},
        "yaxis": {"title": "Pengangguran (X2)"},
        "zaxis": {"title": "Kemiskinan (Y)"},
    },
    width=1000,
    height=750,
    margin={"l": 0, "r": 0, "b": 0, "t": 100},
    legend={"x": 0.02, "y": 0.98},
)

# 15. MENAMPILKAN GRAFIK INTERAKTIF
fig.show()
```
